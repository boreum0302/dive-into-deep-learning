```python
class Vocab:
    def __init__(self, tokens=[], min_freq=0, reserved_tokens=[]):
        if tokens and isinstance(tokens[0], list):
            tokens = [token for line in tokens for token in line]
            '''
            for line in tokens:
                for token in line:
                    token
            '''

        counter = collections.Counter(tokens)
        self.token_freqs = sorted(counter.items(), key=lambda x: x[1], reverse=True)  # 많이 나온 token 순으로 정렬
        self.idx_to_token = sorted(list(set(
            ['<unk>'] + reserved_tokens + [token for token, freq in self.token_freqs if freq >= min_freq]
        )))
        self.token_to_idx = {token: idx for idx, token in enumerate(self.idx_to_token)}

    def __len__(self):
        return len(self.idx_to_token)

    def __getitem__(self, tokens):
        if not isinstance(tokens, (list, tuple)):
            return self.token_to_idx.get(tokens, self.token_to_idx['<unk>'])
        return [self.__getitem__(token) for token in tokens]
        
    def to_tokens(self, indices):
        if hasattr(indices, '__len__') and len(indices) > 1:
            return [self.idx_to_token[int(index)] for index in indices]
        return self.idx_to_token[indices]
```

```python
class Vocab:
    def __init__(self, tokens, min_freq=0, reserved_tokens=None):
        if reserved_tokens is None:
            reserved_tokens = []

        if tokens and isinstance(tokens[0], list):
            tokens = [token for line in tokens for token in line]

        counter = collections.Counter(tokens)
        token_freqs = sorted(counter.items(), key=lambda x: x[1], reverse=True)

        self.idx_to_token = (
            ['<unk>'] +
            reserved_tokens +
            [token for token, freq in token_freqs if freq >= min_freq]
        )
        self.token_to_idx = {token: idx for idx, token in enumerate(self.idx_to_token)}

    def __len__(self):
        return len(self.idx_to_token)

    def token_to_index(self, token: str) -> int:
        return self.token_to_idx.get(token, self.token_to_idx['<unk>'])

    def tokens_to_indices(self, tokens):
        return [self.token_to_index(token) for token in tokens]

    def index_to_token(self, idx: int) -> str:
        return self.idx_to_token[idx]

    def indices_to_tokens(self, indices):
        return [self.idx_to_token[int(idx)] for idx in indices]
```

```python
class TimeMachine:
    def __init__(self, path='./data/timemachine.txt'):
        with open(path, 'r') as f:
            self.raw_text = f.read()

    def _preprocess(self, text):
        return re.sub('[^A-Za-z]+', ' ', text).lower()

    def _tokenize(self, text):
        return list(text)

    def build(self, vocab=None):
        self.text = self._preprocess(self.raw_text)
        self.tokens = self._tokenize(self.text)

        if vocab is None:
            vocab = Vocab(self.tokens)

        corpus = vocab.tokens_to_indices(self.tokens)
        return corpus, vocab
```
